# Run an estimator benchmark

Runs the estimator benchmark of an `estimator-benchmark` exp/run (DESIGN-ESTIMATORS.md, Section 9), and saves `results.pickle` and `results.csv` into its data directory. The results are a long-format table with one row per (environment, sampler, repetition, estimator, field, number of observations, metric).

```python
import pathlib
import sys
sys.path.append("..")
import warnings
warnings.simplefilter("ignore")
from exp_run_config import Config
from estimator_benchmark import run_estimator_benchmark
Config.PROJECTNAME = "WaterBerryFarms"
```

```python
# Reuse the existing experiment directory, creating it only when absent.
# creation_style = "exist-ok"
# Move the existing experiment directory to a timestamped backup, then start fresh.
# creation_style = "version"
# Delete the existing experiment directory, then start fresh.
creation_style = "discard-old"
expruns_path = None
results_path = None
experiment = "estimator-benchmark"
run = "miniberry-30-quick"
# run = "miniberry-30-all"
```

```python
if expruns_path is not None:
    Config().set_exprun_path(pathlib.Path(expruns_path))
if results_path is not None:
    Config().set_results_path(pathlib.Path(results_path))
exp = Config().get_experiment(experiment, run, creation_style=creation_style)
results = run_estimator_benchmark(exp)
last = results[results.observations == results.observations.max()]
print(last[last.metric.isin(["mae", "recall", "brier"]) & (last.field == "tylcv")]
      .pivot_table(index="estimator", columns=["sampler", "metric"], values="value").round(3))
exp.done()
```
