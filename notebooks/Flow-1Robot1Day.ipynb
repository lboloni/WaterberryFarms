# Single-robot benchmark flow

Precompute the required environments, run every selected algorithm, visualize each run, and compare the results.

```python
import pathlib
import sys
sys.path.append("..")
from exp_run_config import Config
from wbf_flow import build_flow_entries, display_flow_report, run_flow, setup_flow
Config.PROJECTNAME = "WaterBerryFarms"
```

```python
flow_name = "single-robot-benchmark"
collection_experiment = "1robot1day"
collection_run = "all-supported"
# Reuse each producer directory, creating it only when absent.
creation_style = "exist-ok"
# Move each existing producer directory to a timestamped backup, then start fresh.
# creation_style = "version"
# Delete each existing producer directory, then start fresh.
# creation_style = "discard-old"
```

```python
expruns_path, results_path, notebooks_path = setup_flow(flow_name, ["environment", "policy", "estimator", "score", "1robot1day"])
entries = build_flow_entries(collection_experiment, collection_run, creation_style)
[(entry["name"], entry["notebook"]) for entry in entries]
```

```python
flow_error = None
try:
    run_flow(entries, expruns_path, results_path, notebooks_path)
except Exception as error:
    flow_error = error
```

```python
final_results_path = pathlib.Path(
    results_path, collection_experiment, collection_run)
preview_files = [
    final_results_path / "score-comparison.pdf",
]
report = display_flow_report(
    entries, results_path, notebooks_path, final_results_path,
    flow_error=flow_error, preview_files=preview_files)
if flow_error is not None:
    raise flow_error
if not report["all-results-present"]:
    raise Exception("Flow finished without all expected stage results.")
```
