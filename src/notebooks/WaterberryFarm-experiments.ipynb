## Put everything together, measure the score for a set of observations


```python
exp_env = precompute_environment(environment_run)
wbf, wbfe = create_wbfe(exp_env)
wbfe.proceed(6)
# observation locations
# for waterberry
# locations = [[1010, 1010], [1500, 1500], [2500, 2500]]
# for miniberry-100
locations = [[10, 10], [30, 30], [50, 50]]
wbfim = WBF_IM_DiskEstimator(wbf.width, wbf.height)
for x, y in locations:
    observation = wbfe.get_observation([x, y, 0])
    # observations passed to a Waterberry estimator carry the robot that made them
    observation["robot"] = "manual"
    wbfim.add_observation(observation)
logging.info("waterberry information model: proceed(1)")
wbfim.proceed(1)
logging.info("done waterberry information model: proceed(1)")
logging.info("starting to calculate score for waterberry information model")
scorecode = WBF_Score_WeightedAsymmetric()
score = scorecode.score(wbfe, wbfim)
logging.info(f"Waterberry information model score: {score}")
# now visualize the environment (top row) and the information model (bottom row)
fig, axes = plt.subplots(2, 4, figsize=(12, 6))
wbfe.geometry.visualize(axes[0, 0])
axes[1, 0].axis("off")
graph_env_im(wbfe, wbfim, ax_env_tylcv=axes[0, 1], ax_im_tylcv=axes[1, 1],
             ax_env_ccr=axes[0, 2], ax_im_ccr=axes[1, 2],
             ax_env_soil=axes[0, 3], ax_im_soil=axes[1, 3])
plt.show()
```
