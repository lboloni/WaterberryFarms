# MRMR 2027: run one simulation

Run one exprun and leave its reusable `results.pickle` artifact.

```python
import pathlib
import sys
sys.path.append("../..")
from exp_run_config import Config
from papers.y2027_mrmr.run_experiment import run_mrmr_experiment
Config.PROJECTNAME = "WaterBerryFarms"
```

```python
# Delete an existing result and run from a clean directory.
creation_style = "discard-old"
# creation_style = "version"
# creation_style = "exist-ok"
expruns_path = None
results_path = None
experiment = "mrmr2027-run"
run = "clustered-mrmr"
# run = "clustered-mrse"
# run = "clustered-mrrw"
# run = "unclustered-mrmr"
# run = "unclustered-mrse"
# run = "unclustered-mrrw"
```

```python
if expruns_path is not None:
    Config().set_exprun_path(pathlib.Path(expruns_path))
if results_path is not None:
    Config().set_results_path(pathlib.Path(results_path))
exp = Config().get_experiment(
    experiment, run, creation_style=creation_style)
results = run_mrmr_experiment(exp)
```

```python
# the bandwidth used by the communication
summary = results["communication-summary"]
print(f"messages: {summary['messages']}, bytes transmitted: {summary['bytes-transmitted']}, "
      f"bytes delivered: {summary['bytes-delivered']}")
print(f"bytes per timestep: mean {summary['mean-bytes-per-timestep']:.1f}, "
      f"max {summary['max-bytes-per-timestep']}")
for message_type, totals in summary["per-type"].items():
    print(f"  {message_type}: {totals}")
for robot_name, cells in results["explored-maps"].items():
    print(f"{robot_name} explored map: {len(cells)} cells")
```
