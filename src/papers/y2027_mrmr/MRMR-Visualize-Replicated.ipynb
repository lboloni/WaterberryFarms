# MRMR 2027: replicated figures

One figure of the replicated comparisons, drawn from the tables of an aggregation (`MRMR-Aggregate.ipynb`): the means with their 95% confidence intervals, and the individual replications or their range (DESIGN-MultiSeedEvaluation.md, Section 7). The `figure-kind` of the figure exp/run selects the figure.

```python
import pathlib
import sys
sys.path.append("../..")
from exp_run_config import Config
Config.PROJECTNAME = "WaterBerryFarms"
from papers.y2027_mrmr.mrmr_aggregate import load_tables
from papers.y2027_mrmr.mrmr_graphics import (show_replicated_communication, show_replicated_comparison,
    show_replicated_paired, show_replicated_per_role, show_replicated_ranges, show_replicated_series)
```

```python
creation_style = "exist-ok"
# creation_style = "version"
# creation_style = "discard-old"
expruns_path = None
results_path = None
experiment = "mrmr2027-figure"
run = "replicated-comparison"
# run = "replicated-ranges"
# run = "replicated-voi-over-time"
# run = "replicated-paired"
# run = "replicated-communication-cost"
# run = "replicated-per-role"
```

```python
if expruns_path is not None:
    Config().set_exprun_path(pathlib.Path(expruns_path))
if results_path is not None:
    Config().set_results_path(pathlib.Path(results_path))
figure_exp = Config().get_experiment(experiment, run, creation_style=creation_style)
source = Config().get_experiment(figure_exp["source-experiment"], figure_exp["source-runs"][0], create_data_dir=False)
tables = load_tables(source["data_dir"])
if "summary" not in tables:
    raise FileNotFoundError(f"No aggregated tables in {source['data_dir']}")
```

```python
kind = figure_exp["figure-kind"]
labels = figure_exp["labels"]
colors = ["#CC6666", "#9999CC", "#66CC99"]
output = figure_exp["output-filename"]
if kind == "comparison":
    show_replicated_comparison(figure_exp, tables, figure_exp["scenarios"], figure_exp["metrics"],
                               figure_exp["approaches"], labels, colors, output)
elif kind == "ranges":
    show_replicated_ranges(figure_exp, tables, figure_exp["scenarios"], figure_exp["metrics"],
                           figure_exp["approaches"], labels, colors, output)
elif kind == "series":
    show_replicated_series(figure_exp, tables, figure_exp["scenarios"], figure_exp["metric"],
                           figure_exp["approaches"], labels, colors, output)
elif kind == "paired":
    show_replicated_paired(figure_exp, tables, figure_exp["scenarios"], figure_exp["metric"], labels, output)
elif kind == "communication":
    show_replicated_communication(figure_exp, tables, figure_exp["scenarios"], figure_exp["approach"],
                                  labels, ["#66CC99", "#9999CC"], output)
elif kind == "per-role":
    show_replicated_per_role(figure_exp, tables, figure_exp["scenarios"], figure_exp["metric"], labels,
                             ["#CC6666", "#9999CC", "#66CC99", "#E69F00"], output)
else:
    raise Exception(f"Unknown figure kind {kind}")
figure_exp.done()
```
