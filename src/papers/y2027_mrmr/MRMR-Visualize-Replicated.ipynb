# MRMR 2027: replicated figures

One figure of the replicated comparisons, drawn from the tables of an aggregation (`MRMR-Aggregate.ipynb`): the means with their 95% confidence intervals, and the individual replications or their range (DESIGN-MultiSeedEvaluation.md, Section 7). The `figure-kind` of the figure exp/run selects the figure.

```python
import pathlib
import sys
sys.path.append("../..")
from exp_run_config import Config
Config.PROJECTNAME = "WaterBerryFarms"
from exprunflow.aggregate import load_tables
from exprunflow.plots import plot_bars, plot_paired, plot_ranges, plot_series
from papers.y2027_mrmr.mrmr_graphics import METRIC_LABELS, show_replicated_communication, show_replicated_per_role
```

```python
creation_style = "exist-ok"
# creation_style = "version"
# creation_style = "discard-old"
expruns_path = None
results_path = None
experiment = "mrmr2027-figure"
run = "replicated-communication-cost-100"
# run = "replicated-communication-cost-200"
# run = "replicated-comparison-100"
# run = "replicated-comparison-200"
# run = "replicated-paired-100"
# run = "replicated-paired-200"
# run = "replicated-per-role-100"
# run = "replicated-per-role-200"
# run = "replicated-ranges-100"
# run = "replicated-ranges-200"
# run = "replicated-voi-over-time-100"
# run = "replicated-voi-over-time-200"
```

```python
if expruns_path is not None:
    Config().set_exprun_path(pathlib.Path(expruns_path))
if results_path is not None:
    Config().set_results_path(pathlib.Path(results_path))
figure_exp = Config().get_experiment(experiment, run, creation_style=creation_style)
source = Config().get_experiment(figure_exp["source-experiment"], figure_exp["source-runs"][0], create_data_dir=False)
tables = load_tables(source["data_dir"])
if "summary" not in tables:
    raise FileNotFoundError(f"No aggregated tables in {source['data_dir']}")
```

```python
kind = figure_exp["figure-kind"]
labels = figure_exp["labels"]
where = figure_exp.get("where")
colors = ["#CC6666", "#9999CC", "#66CC99"]
output = pathlib.Path(figure_exp.data_dir(), figure_exp["output-filename"])
common = {"lookup": labels, "where": where, "metric_labels": METRIC_LABELS}
if kind == "bars":
    plot_bars(tables, figure_exp["metrics"], "scenario", "approach", output, x_values=figure_exp["scenarios"],
              color_values=figure_exp["approaches"], colors=colors, **common)
elif kind == "ranges":
    plot_ranges(tables, figure_exp["metrics"], "scenario", "approach", output, x_values=figure_exp["scenarios"],
                color_values=figure_exp["approaches"], colors=colors, **common)
elif kind == "series":
    plot_series(tables, figure_exp["metric"], "scenario", "approach", output, panel_values=figure_exp["scenarios"],
                color_values=figure_exp["approaches"], colors=colors, **common)
elif kind == "paired":
    plot_paired(tables, figure_exp["metric"], "scenario", "approach", output, panel_values=figure_exp["scenarios"],
                **common)
elif kind == "communication":
    show_replicated_communication(tables, figure_exp["scenarios"], figure_exp["approach"], labels,
                                  ["#66CC99", "#9999CC"], output, where=where)
elif kind == "per-role":
    show_replicated_per_role(tables, figure_exp["scenarios"], figure_exp["metric"], labels,
                             ["#CC6666", "#9999CC", "#66CC99", "#E69F00"], output, where=where)
else:
    raise Exception(f"Unknown figure kind {kind}")
figure_exp.done()
```
