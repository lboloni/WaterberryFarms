# MRMR 2027: optimal exploration-package path

Generate the configured optimal-EP-path illustration without simulation results.

```python
import pathlib
import sys
sys.path.append("../..")
import matplotlib.patches as patches
import matplotlib.pyplot as plt
from exp_run_config import Config
from papers.y2027_mrmr.exploration_package import ExplorationPackage, ExplorationPackageSet
from papers.y2027_mrmr.mrmr_graphics import save_figure
from path_generators import euclidean_distance
Config.PROJECTNAME = "WaterBerryFarms"
```

```python
creation_style = "exist-ok"
# creation_style = "version"
# creation_style = "discard-old"
expruns_path = None
results_path = None
experiment = "mrmr2027-figure"
run = "optimal-ep-path"
```

```python
if expruns_path is not None:
    Config().set_exprun_path(pathlib.Path(expruns_path))
if results_path is not None:
    Config().set_results_path(pathlib.Path(results_path))
figure_exp = Config().get_experiment(
    experiment, run, creation_style=creation_style)
packages = ExplorationPackageSet()
for values in figure_exp["exploration-packages"]:
    packages.add_ep(ExplorationPackage(
        values["x-min"], values["x-max"],
        values["y-min"], values["y-max"], values["step"]))
path, _ = packages.find_shortest_path_ep(
    start=figure_exp["start"], end=figure_exp["end"],
    max_evaluations=figure_exp["max-evaluations"])
```

```python
fig, ax = plt.subplots(1, 1, figsize=(5, 5))
ax.invert_xaxis()
ax.set_xlim(0, 100)
ax.set_ylim(0, 100)
previous_point = None
for point in path:
    if previous_point is not None:
        arrowstyle = (
            "->" if euclidean_distance(point, previous_point) > 5
            else "-")
        ax.add_patch(patches.FancyArrowPatch(
            previous_point, point, arrowstyle=arrowstyle,
            mutation_scale=10, color="#E69F00",
            shrinkA=0, shrinkB=0))
    previous_point = point
save_figure(fig, pathlib.Path(
    figure_exp["data_dir"], figure_exp["output-filename"]))
plt.close(fig)
figure_exp.done()
```
