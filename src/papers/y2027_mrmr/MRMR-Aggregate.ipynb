# MRMR 2027: aggregate the replications

Collect the `metrics.json` of the replications declared by an aggregate exp/run (its base runs, map seeds and behavior seeds), and write the tidy tables and their statistics into its data directory: `replications.csv`, `robots.csv`, `series.csv`, `summary.csv`, `summary-series.csv`, `paired.csv` and `variance.csv` (DESIGN-MultiSeedEvaluation.md, Section 6). The replicated figures draw only from these tables.

```python
import pathlib
import sys
sys.path.append("../..")
import pandas as pd
from exp_run_config import Config
Config.PROJECTNAME = "WaterBerryFarms"
from papers.y2027_mrmr.mrmr_aggregate import aggregate_tables, load_replications, save_tables
```

```python
creation_style = "exist-ok"
# creation_style = "version"
# creation_style = "discard-old"
expruns_path = None
results_path = None
experiment = "mrmr2027-aggregate"
run = "icc-2027-replicated"
```

```python
if expruns_path is not None:
    Config().set_exprun_path(pathlib.Path(expruns_path))
if results_path is not None:
    Config().set_results_path(pathlib.Path(results_path))
exp = Config().get_experiment(experiment, run, creation_style=creation_style)
metrics = load_replications(exp)
tables = aggregate_tables(metrics, exp["approaches"])
save_tables(tables, exp.data_dir())
print(f"Aggregated {len(metrics)} replications into {exp.data_dir()}")
```
