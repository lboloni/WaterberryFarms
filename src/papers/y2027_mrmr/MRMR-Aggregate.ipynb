# MRMR 2027: aggregate the replications

Collect the `metrics.json` of the replications declared by an aggregate exp/run (its base runs and factors), and write the tidy tables and their statistics into its data directory, with the ExpRunFlow library (`exprunflow.aggregate`; ExpRunFlow `docs/DESIGN-Replication.md`, and DESIGN-MultiSeedEvaluation.md): `replications.csv`, `entities.csv`, `series.csv`, `summary.csv`, `summary-series.csv`, `paired.csv`, `variance.csv` and `hosts.csv`. The replicated figures draw only from these tables.

```python
import pathlib
import sys
sys.path.append("../..")
import pandas as pd
from exp_run_config import Config
Config.PROJECTNAME = "WaterBerryFarms"
from exprunflow.aggregate import aggregate_tables, load_replications, save_tables
```

```python
creation_style = "exist-ok"
# creation_style = "version"
# creation_style = "discard-old"
expruns_path = None
results_path = None
experiment = "mrmr2027-aggregate"
run = "icc-2027-replicated"
```

```python
if expruns_path is not None:
    Config().set_exprun_path(pathlib.Path(expruns_path))
if results_path is not None:
    Config().set_results_path(pathlib.Path(results_path))
exp = Config().get_experiment(experiment, run, creation_style=creation_style)
metrics = load_replications(exp)
tables = aggregate_tables(metrics, exp)
save_tables(tables, exp.data_dir())
print(f"Aggregated {len(metrics)} replications into {exp.data_dir()}")
tables["hosts"]
```

```python
summary = tables["summary"]
columns = ["map-size", "scenario", "approach", "metric", "n", "mean", "ci95_low_clusters", "ci95_high_clusters", "min", "max"]
summary[summary.metric.isin(["voi-absolute", "diseased-found", "bytes-transmitted", "eps-awarded", "computation-seconds"])][columns]
```
