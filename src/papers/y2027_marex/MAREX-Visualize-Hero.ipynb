# MAREX 2027: hero figure

Compile the TikZ schematic of the hero figure (the field, the Pioneer and Contractor drones, the exploration packages and the EP market messages) with xelatex, without simulation results.

```python
import pathlib
import shutil
import subprocess
import sys
import tempfile
sys.path.append("../..")
from exp_run_config import Config
Config.PROJECTNAME = "WaterBerryFarms"
```

```python
creation_style = "exist-ok"
# creation_style = "version"
# creation_style = "discard-old"
expruns_path = None
results_path = None
experiment = "marex2027-figure"
run = "hero"
```

```python
if expruns_path is not None:
    Config().set_exprun_path(pathlib.Path(expruns_path))
if results_path is not None:
    Config().set_results_path(pathlib.Path(results_path))
figure_exp = Config().get_experiment(
    experiment, run, creation_style=creation_style)
source = pathlib.Path(figure_exp["tikz-source"]).resolve()
```

```python
# compile in a scratch directory, so that the auxiliary files of xelatex stay out of the results
output = pathlib.Path(figure_exp["data_dir"], figure_exp["output-filename"])
with tempfile.TemporaryDirectory() as build:
    subprocess.run(["xelatex", "-interaction=nonstopmode", "-halt-on-error",
                    f"-output-directory={build}", str(source)], check=True,
                   cwd=source.parent, stdout=subprocess.DEVNULL)
    shutil.copy2(pathlib.Path(build, source.stem + ".pdf"), output)
# the PNG is the inline preview of the flow report, as in save_figure
subprocess.run(["pdftoppm", "-r", "150", "-png", "-singlefile", str(output),
                str(output.with_suffix(""))], check=True)
print(f"Done saving to {output} and {output.with_suffix('.png')}")
```

```python
figure_exp.done()
```
