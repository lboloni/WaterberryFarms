# MAREX 2027: per-agent VoI

Create one bar chart of the VoI credited to each agent of a saved run.

```python
import gzip
import pathlib
import pickle
import sys
sys.path.append("../..")
from exp_run_config import Config
Config.PROJECTNAME = "WaterBerryFarms"
from papers.y2027_marex.marex_graphics import show_agentwise_voi
```

```python
creation_style = "exist-ok"
# creation_style = "version"
# creation_style = "discard-old"
expruns_path = None
results_path = None
experiment = "marex2027-figure"
run = "agent-voi-clustered-marex"
# run = "agent-voi-clustered-mrrw"
# run = "agent-voi-unclustered-marex"
# run = "agent-voi-unclustered-mrrw"
```

```python
if expruns_path is not None:
    Config().set_exprun_path(pathlib.Path(expruns_path))
if results_path is not None:
    Config().set_results_path(pathlib.Path(results_path))
figure_exp = Config().get_experiment(
    experiment, run, creation_style=creation_style)
source_run = figure_exp["source-runs"][0]
source_exp = Config().get_experiment(
    figure_exp["source-experiment"], source_run,
    create_data_dir=False)
results_file = pathlib.Path(source_exp["data_dir"], "results.pickle")
if not results_file.exists():
    raise FileNotFoundError(results_file)
with gzip.open(results_file, "rb") as handle:
    results = pickle.load(handle)
```

```python
robot_colors = ["#E69F00", "#56B4E9", "#009E73"]
show_agentwise_voi(
    figure_exp, source_run, results, robot_colors,
    variant=figure_exp["voi-variant"],
    output_filename=figure_exp["output-filename"])
figure_exp.done()
```
