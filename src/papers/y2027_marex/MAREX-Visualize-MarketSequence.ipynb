# MAREX 2027: the EP market as a message sequence chart

The negotiation of one exploration package in a saved MAREX run, drawn from the messages the robots actually sent (the traffic records of the communication medium): the detection streak of the Pioneer, the offer broadcast, the feasibility checks and bids of the Contractors, the award, the replanning and exploration of the winner, and the completion report, each message with its size in bytes.

```python
import gzip
import pathlib
import pickle
import sys
sys.path.append("../..")
from exp_run_config import Config
Config.PROJECTNAME = "WaterBerryFarms"
from papers.y2027_marex.marex_graphics import show_market_sequence
```

```python
creation_style = "exist-ok"
# creation_style = "version"
# creation_style = "discard-old"
expruns_path = None
results_path = None
experiment = "marex2027-figure"
run = "market-sequence"
```

```python
if expruns_path is not None:
    Config().set_exprun_path(pathlib.Path(expruns_path))
if results_path is not None:
    Config().set_results_path(pathlib.Path(results_path))
figure_exp = Config().get_experiment(
    experiment, run, creation_style=creation_style)
source_exp = Config().get_experiment(
    figure_exp["source-experiment"], figure_exp["source-runs"][0],
    create_data_dir=False)
results_file = pathlib.Path(source_exp["data_dir"], "results.pickle")
if not results_file.exists():
    raise FileNotFoundError(results_file)
with gzip.open(results_file, "rb") as handle:
    results = pickle.load(handle)
```

```python
lifecycle = show_market_sequence(
    figure_exp, results, offer_id=figure_exp["offer-id"],
    output_filename=figure_exp["output-filename"])
print(f"Offer {lifecycle['offer-id']}: {len(lifecycle['bids'])} bids, awarded to "
      f"{lifecycle['award']['destination'] if lifecycle['award'] else None}")
figure_exp.done()
```
